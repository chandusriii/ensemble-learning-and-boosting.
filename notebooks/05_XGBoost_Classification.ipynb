# XGBoost Classification

XGBoost is an optimized gradient-boosted tree library. This notebook demonstrates binary classification with a synthetic dataset.

```python
import xgboost as xgb
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

X, y = make_classification(
    n_samples=500, n_features=20, n_informative=10,
    n_redundant=2, random_state=42
)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

model = xgb.XGBClassifier(
    n_estimators=100, max_depth=3, learning_rate=0.1,
    eval_metric="logloss", random_state=42
)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification report:\n", classification_report(y_test, y_pred))
```

## Parameters to explore

`n_estimators` controls the number of boosting rounds, `max_depth` controls tree complexity, and `learning_rate` scales each round's contribution. Further extensions could compare regularization, subsampling, early stopping, and an XGBoost regressor.
