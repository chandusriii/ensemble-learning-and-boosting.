# Gradient Boosting: Regression and Classification

This notebook separates regression and classification experiments and reports task-appropriate metrics.

## 1. Gradient Boosting Regressor

```python
from sklearn.datasets import make_regression, make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.metrics import accuracy_score, classification_report
from sklearn.ensemble import GradientBoostingRegressor, GradientBoostingClassifier
import numpy as np

X_reg, y_reg = make_regression(
    n_samples=1000, n_features=10, noise=20, random_state=42
)
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

gbr = GradientBoostingRegressor(
    n_estimators=200, learning_rate=0.05,
    max_depth=3, subsample=0.8, random_state=42
)
gbr.fit(X_train_reg, y_train_reg)
reg_pred = gbr.predict(X_test_reg)

print("MAE:", mean_absolute_error(y_test_reg, reg_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test_reg, reg_pred)))
print("R²:", r2_score(y_test_reg, reg_pred))
```

## 2. Gradient Boosting Classifier

```python
X_cls, y_cls = make_classification(
    n_samples=500, n_features=20, n_informative=10,
    n_redundant=2, random_state=42
)
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls, y_cls, test_size=0.3, random_state=42, stratify=y_cls
)

gbc = GradientBoostingClassifier(
    learning_rate=0.1, n_estimators=150,
    max_depth=3, random_state=42
)
gbc.fit(X_train_cls, y_train_cls)
cls_pred = gbc.predict(X_test_cls)

print("Accuracy:", accuracy_score(y_test_cls, cls_pred))
print("\nClassification report:\n", classification_report(y_test_cls, cls_pred))
```

## Key parameters

- `n_estimators`: number of sequential boosting stages.
- `learning_rate`: scales each stage's contribution.
- `max_depth`: controls individual tree complexity.
- `subsample`: fraction of training data used per stage for the regressor above.

Tune these on the training set with cross-validation; reserve the held-out test set for final reporting.
